# OCR Form Checker

This project consists of two parts invovled in a form checking process. The first part is based on signature recognition, which should match one signature to another. The idea for this has not been expanded on yet, as I have not decided whether I intend to use ML or mathematical methods.

The second part invovles reading a form with data in, and checking whether information matches what is expected based on a database. This would then automate this checking process, any that match completely, are dated correctly, and have a signature that matches (this may not need to be true in the real use case) then they pass and do not qualify for being reviewed manually. If however a form does not meet any of these criteria, then they are held and will need to be reviewed manually.

This notebook in particular is responsible for performing the appropriate ETL for each part of this project. It may however be split into multiple notebooks in the future, should the project need it to be for readability.
